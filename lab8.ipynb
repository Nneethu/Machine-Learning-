```python
# ============================================================
# LAB 8 - CATEGORICAL NAIVE BAYES CLASSIFIER
# Play Tennis Dataset
# ============================================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder
from sklearn.naive_bayes import CategoricalNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)
```

```python
# ============================================================
# Step 1: Load the dataset
# ============================================================

# Change the file name if required
file_path = r"C:\Users\Neethu\Documents\ML\Lab 8.xlsx"

df = pd.read_excel(file_path)

print("Original Dataset")
print(df)

print("\nDataset Shape:", df.shape)
```

```python
# ============================================================
# Step 2: Separate input features and target variable
# ============================================================

X = df[['Outlook', 'Temperature', 'Humidity', 'Wind']]
y = df['Play Tennis']

print("\nInput Features:")
print(X.head())

print("\nTarget Variable:")
print(y.head())
```

```python
# ============================================================
# Step 3: Divide the dataset into training and testing sets
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining Data Shape:", X_train.shape)
print("Testing Data Shape:", X_test.shape)
```

```python
# ============================================================
# Step 4: Convert categorical values into numerical values
# Using OrdinalEncoder for Categorical Naive Bayes
# ============================================================

encoder = OrdinalEncoder()

X_train_encoded = encoder.fit_transform(X_train)
X_test_encoded = encoder.transform(X_test)

print("\nEncoded Training Data:")
print(X_train_encoded)

print("\nEncoded Testing Data:")
print(X_test_encoded)
```

```python
# ============================================================
# Step 5: Train Categorical Naive Bayes model
# ============================================================

nb_model = CategoricalNB()

nb_model.fit(X_train_encoded, y_train)
```

```python
# ============================================================
# Step 6: Predict class labels for test dataset
# ============================================================

y_pred_nb = nb_model.predict(X_test_encoded)

print("\nNaive Bayes Predictions:")
print(y_pred_nb)
```

```python
# ============================================================
# Step 7: Calculate accuracy
# ============================================================

nb_accuracy = accuracy_score(y_test, y_pred_nb)

print("\n======================================")
print("Categorical Naive Bayes Performance")
print("======================================")

print("Accuracy:", nb_accuracy)
```

```python
# ============================================================
# Step 8: Display confusion matrix
# ============================================================

cm_nb = confusion_matrix(
    y_test,
    y_pred_nb,
    labels=['No', 'Yes']
)

print("\nConfusion Matrix:")
print(cm_nb)
```

```python
# ============================================================
# Step 9: Display classification report
# ============================================================

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred_nb,
        labels=['No', 'Yes']
    )
)
```

```python
# ============================================================
# Step 10: Predict for new weather condition
# ============================================================

new_data = pd.DataFrame({
    'Outlook': ['Sunny'],
    'Temperature': ['Cool'],
    'Humidity': ['High'],
    'Wind': ['Strong']
})

print("\n======================================")
print("New Weather Condition")
print("======================================")

print(new_data)

# Encode new data using the same encoder
new_data_encoded = encoder.transform(new_data)

# Predict class
new_prediction = nb_model.predict(new_data_encoded)

# Predict class probabilities
new_probability = nb_model.predict_proba(new_data_encoded)

print("\nPredicted Class:")
print(new_prediction[0])

print("\nClass Probabilities:")

for class_name, probability in zip(
    nb_model.classes_,
    new_probability[0]
):
    print(class_name, ":", round(probability, 4))
```

```python
# ============================================================
# Step 11: Compare with Decision Tree and Logistic Regression
# ============================================================

# One-hot encoding is used for Decision Tree and
# Logistic Regression because these models work well
# with one-hot encoded categorical features.

onehot_encoder = OneHotEncoder(handle_unknown='ignore')

X_train_onehot = onehot_encoder.fit_transform(X_train)
X_test_onehot = onehot_encoder.transform(X_test)

# ------------------------------------------------------------
# Decision Tree
# ------------------------------------------------------------

dt_model = DecisionTreeClassifier(
    random_state=42
)

dt_model.fit(X_train_onehot, y_train)

y_pred_dt = dt_model.predict(X_test_onehot)

dt_accuracy = accuracy_score(
    y_test,
    y_pred_dt
)

# ------------------------------------------------------------
# Logistic Regression
# ------------------------------------------------------------

lr_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

lr_model.fit(X_train_onehot, y_train)

y_pred_lr = lr_model.predict(X_test_onehot)

lr_accuracy = accuracy_score(
    y_test,
    y_pred_lr
)
```

```python
# ============================================================
# Step 12: Display comparison
# ============================================================

print("\n======================================")
print("MODEL PERFORMANCE COMPARISON")
print("======================================")

print("Categorical Naive Bayes Accuracy :",
      round(nb_accuracy, 4))

print("Decision Tree Accuracy           :",
      round(dt_accuracy, 4))

print("Logistic Regression Accuracy     :",
      round(lr_accuracy, 4))
```

```python
# ============================================================
# Step 13: Comparison Table
# ============================================================

comparison = pd.DataFrame({
    'Model': [
        'Categorical Naive Bayes',
        'Decision Tree',
        'Logistic Regression'
    ],
    'Accuracy': [
        nb_accuracy,
        dt_accuracy,
        lr_accuracy
    ]
})

print("\nModel Comparison:")
print(comparison)
```

```python
# ============================================================
# Step 14: Display confusion matrices for comparison
# ============================================================

print("\nNaive Bayes Confusion Matrix:")
print(
    confusion_matrix(
        y_test,
        y_pred_nb,
        labels=['No', 'Yes']
    )
)

print("\nDecision Tree Confusion Matrix:")
print(
    confusion_matrix(
        y_test,
        y_pred_dt,
        labels=['No', 'Yes']
    )
)

print("\nLogistic Regression Confusion Matrix:")
print(
    confusion_matrix(
        y_test,
        y_pred_lr,
        labels=['No', 'Yes']
    )
)
```
