```python
print("Observation:")
print("The datasets contain missing values and possible duplicate records.")
print("The air-quality dataset contains daily city-level measurements.")
print("The crop dataset contains district-level crop production records.")
print("These issues must be handled before combining or analysing the datasets.")
```

```python
#TASK 2 — Missing Values
# Missing values BEFORE treatment

print("AIR QUALITY - BEFORE")
print(air.isnull().sum())

print("\nCROP DATA - BEFORE")
print(crop.isnull().sum())
```

```python
# Convert date columns

air["Date"] = pd.to_datetime(air["Date"], errors="coerce")

# Clean crop text columns
crop["State_Name"] = crop["State_Name"].str.strip()
crop["District_Name"] = crop["District_Name"].str.strip()
crop["Season"] = crop["Season"].str.strip()
crop["Crop"] = crop["Crop"].str.strip()


# Numeric columns -> median
air_numeric = air.select_dtypes(include=np.number).columns

for col in air_numeric:
    air[col] = air[col].fillna(air[col].median())


# AQI Bucket -> mode
if "AQI_Bucket" in air.columns:
    air["AQI_Bucket"] = air["AQI_Bucket"].fillna(
        air["AQI_Bucket"].mode()[0]
    )


# Crop Production -> median
crop["Production"] = crop["Production"].fillna(
    crop["Production"].median()
)
```

```python
print("AIR QUALITY - AFTER")
print(air.isnull().sum())

print("\nCROP DATA - AFTER")
print(crop.isnull().sum())
```

```python
#TASK 3 — State Names and Duplicates
# Check duplicate records

print("Air duplicates before:", air.duplicated().sum())
print("Crop duplicates before:", crop.duplicated().sum())

# Remove duplicates

air = air.drop_duplicates()
crop = crop.drop_duplicates()

print("\nAir duplicates after:", air.duplicated().sum())
print("Crop duplicates after:", crop.duplicated().sum())
```

```python
# Clean crop state names

crop["State_Name"] = (
    crop["State_Name"]
    .str.strip()
    .str.upper()
)

# Display state names
print("State names in crop dataset:")
print(sorted(crop["State_Name"].unique()))
```

```python
#TASK 4 — AQI Distribution
plt.figure(figsize=(10,5))

plt.hist(air["AQI"], bins=30)

plt.xlabel("AQI")
plt.ylabel("Number of Records")
plt.title("Distribution of AQI")

plt.show()
```

```python
#TASK 5 — Outlier Treatment
# Find AQI outliers using IQR

Q1 = air["AQI"].quantile(0.25)
Q3 = air["AQI"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = air[
    (air["AQI"] < lower) |
    (air["AQI"] > upper)
]

print("Lower limit:", lower)
print("Upper limit:", upper)
print("Number of outliers:", len(outliers))
```

```python
air_before = air["AQI"].copy()

air["AQI"] = air["AQI"].clip(lower=lower, upper=upper)

print("Outliers handled using IQR capping.")
```

```python
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.boxplot(air_before)
plt.title("AQI Before Treatment")
plt.ylabel("AQI")

plt.subplot(1,2,2)
plt.boxplot(air["AQI"])
plt.title("AQI After Treatment")
plt.ylabel("AQI")

plt.tight_layout()
plt.show()
```

```python
#TASK 6 — AQI Over Time
# Extract year

air["Year"] = air["Date"].dt.year

yearly_aqi = air.groupby("Year")["AQI"].mean()

print(yearly_aqi)
```

```python
plt.figure(figsize=(10,5))

plt.plot(yearly_aqi.index, yearly_aqi.values, marker="o")

plt.xlabel("Year")
plt.ylabel("Average AQI")
plt.title("Average AQI Over Time")

plt.grid()
plt.show()
```

```python
highest_year = yearly_aqi.idxmax()
lowest_year = yearly_aqi.idxmin()

print("Most polluted year:", highest_year)
print("Average AQI:", yearly_aqi.max())

print("\nLeast polluted year:", lowest_year)
print("Average AQI:", yearly_aqi.min())
```

```python
#TASK 7 — Seasonal AQI Pattern
# Extract month

air["Month"] = air["Date"].dt.month

monthly_aqi = air.groupby("Month")["AQI"].mean()

plt.figure(figsize=(10,5))

plt.plot(monthly_aqi.index, monthly_aqi.values, marker="o")

plt.xlabel("Month")
plt.ylabel("Average AQI")
plt.title("Average AQI by Month")

plt.xticks(range(1,13))
plt.grid()

plt.show()
```

```python
def get_season(month):
    if month in [10, 11, 12]:
        return "Oct-Dec"
    elif month in [1, 2, 3]:
        return "Jan-Mar"
    elif month in [4, 5, 6]:
        return "Apr-Jun"
    else:
        return "Jul-Sep"

air["Season"] = air["Month"].apply(get_season)

season_aqi = air.groupby("Season")["AQI"].mean()

print(season_aqi)
```

```python
plt.figure(figsize=(8,5))

season_aqi.plot(kind="bar")

plt.xlabel("Season")
plt.ylabel("Average AQI")
plt.title("Average AQI by Season")

plt.xticks(rotation=0)
plt.show()
```

```python
highest_season = season_aqi.idxmax()

print("Highest AQI season:", highest_season)
print("Average AQI:", season_aqi.max())

print("\nThe October-December claim should be accepted only if this season has the highest AQI.")
```

```python
#TASK 8 — Combine Air Quality and Crop Data
state_year_aqi = (
    air.groupby(["State", "Year"])["AQI"]
    .mean()
    .reset_index()
)

state_year_aqi.columns = ["State", "Year", "Average_AQI"]

print(state_year_aqi.head())
```

```python
crop["State_Name"] = crop["State_Name"].str.upper()

crop_year = (
    crop.groupby(["State_Name", "Crop_Year"])["Production"]
    .sum()
    .reset_index()
)

crop_year.columns = ["State", "Year", "Production"]

print(crop_year.head())
```

```python
combined = pd.merge(
    state_year_aqi,
    crop_year,
    on=["State", "Year"],
    how="inner"
)

print("Combined dataset shape:", combined.shape)

display(combined.head())
```

```python
print("Air quality years:",
      air["Year"].min(), "to", air["Year"].max())

print("Crop years:",
      crop["Crop_Year"].min(), "to", crop["Crop_Year"].max())

print("\nCommon years:")
print(sorted(set(air["Year"].unique()) &
             set(crop["Crop_Year"].unique())))
```

```python
#Correlation Analysis
correlation = combined[
    ["Average_AQI", "Production"]
].corr()

print(correlation)
```

```python
plt.figure(figsize=(6,5))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Between AQI and Crop Production")

plt.show()
```
